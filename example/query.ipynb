```python
from os import getenv

from dotenv import load_dotenv
from voxrow.query import Query
```

> Pastikan _file_ [`query.env`](query.env) sudah diisi dengan kredensial _database_ yang anda miliki.

```python
load_dotenv('query.env')
DB_HOST=getenv('DB_HOST')
DB_PORT=getenv('DB_PORT')
DB_NAME=getenv('DB_NAME')
DB_USER=getenv('DB_USER')
DB_PASSWORD=getenv('DB_PASSWORD')
```

```python
query = Query(f'postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}')
```

```python
query(
    '''
        SELECT
            *
        FROM
            {{schema_name}}.{{table_name}}
        WHERE
            schemaname = :schema_name
        LIMIT 1
        ;
    ''',
    schema_name='pg_catalog',
    table_name='pg_tables'
)
```

```python
query(
    'query.sql',
    schema_name='pg_catalog',
    table_name='pg_tables'
)
```
